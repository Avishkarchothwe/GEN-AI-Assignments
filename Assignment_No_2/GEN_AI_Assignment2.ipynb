```python
import requests
import json
API_KEY = "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"
```

```python
def ask_llm(prompt):

    url = "https://openrouter.ai/api/v1/chat/completions"

    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json"
    }

    data = {
        "model": "deepseek/deepseek-chat",
        "messages": [
            {"role": "user", "content": prompt}
        ]
    }

    response = requests.post(url, headers=headers, data=json.dumps(data))

    result = response.json()

    return result["choices"][0]["message"]["content"]
```

1. Zero Shot Prompting

```python
zero_prompt = """
Classify the sentiment of this sentence as Positive, Negative, or Neutral.

Sentence: The movie was fantastic and inspiring.
"""

result = ask_llm(zero_prompt)

print("ZERO SHOT RESULT\n")
print(result)
```

2. Few Shot Prompting

```python
few_prompt = """
Classify the sentiment.

Sentence: I love this product -> Positive
Sentence: The service was terrible -> Negative
Sentence: The book was okay -> Neutral

Sentence: The movie was fantastic and inspiring ->
"""

result = ask_llm(few_prompt)

print("FEW SHOT RESULT\n")
print(result)
```

3. Interview Prompting

```python
questions = [
    "What is Machine Learning?",
    "What are the types of Machine Learning?",
    "Give real-world applications of Machine Learning."
]

print("INTERVIEW APPROACH\n")

for q in questions:
    ans = ask_llm(q)
    print("Question:", q)
    print("Answer:", ans)
    print("-----------------------")
```

4. Chain of Thought Prompting

```python
cot_prompt = """
Solve the following problem step by step.

If a train travels 60 km in 1 hour, how far will it travel in 3 hours?
Explain your reasoning before giving the final answer.
"""

result = ask_llm(cot_prompt)

print("CHAIN OF THOUGHT RESULT:\n")
print(result)
```

5. Tree of Thought Prompting

```python
tot_prompt = """
Solve the puzzle using Tree of Thought reasoning.

Problem:
A farmer has a wolf, goat, and cabbage and must cross a river.
He can only carry one item at a time.

Steps:
1. Generate multiple strategies.
2. Evaluate each strategy.
3. Choose the best solution and explain it.
"""

result = ask_llm(tot_prompt)

print("TREE OF THOUGHT RESULT:\n")
print(result)
```
