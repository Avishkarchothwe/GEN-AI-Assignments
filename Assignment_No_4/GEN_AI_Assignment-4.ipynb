4. Text Preprocessing Functions

```python
def preprocess_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Tokenization
    tokens = nltk.word_tokenize(text)

    # Stopword removal
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]

    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    return " ".join(tokens)
```

5. Load Pre-trained Question Answering Model

```python
qa_model = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad"
)
```

6. Question Answering Function

```python
def get_answer(question, context):

    result = qa_model(
        question=question,
        context=context
    )

    answer = result['answer']
    score = result['score']

    return answer, score
```

7. Chatbot Interface

```python
def chatbot():

    print("====================================")
    print("Question Answering Chatbot")
    print("Type 'exit' to stop the chatbot")
    print("====================================")

    while True:

        question = input("\nUser: ")

        if question.lower() == "exit":
            print("Chatbot: Goodbye!")
            break

        answer, score = get_answer(question, context)

        print("Chatbot:", answer)
        print("Confidence Score:", round(score,3))
```

8. Run the Chatbot

```python
if __name__ == "__main__":

    chatbot()
```
